Atividade: Criar uma coluna nova `price` e completar usando o modelo de redes neurais do Scikit-Learn (ANN), todos os conjuntos de leve devem ter preço.

Quais as colunas importantes para prever o preço??

```python
---
## Solução da atividade Só **adiciona** a coluna `price` em `lego_df`. As colunas originais (incluindo `US_retailPrice`) não são alteradas; a imputação de faltantes das *features* acontece dentro do pipeline do modelo.
```

```python
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score

# NOVA coluna: começa igual ao preço original (NaN onde não há preço)
lego_df['price'] = lego_df['US_retailPrice']
sem_preco = lego_df['price'].isna()          # máscara (não vira coluna)
print('Com preço:', (~sem_preco).sum(), '| Sem preço:', sem_preco.sum())

num_cols = ['year', 'pieces', 'minifigs', 'agerange_min']
cat_cols = ['theme', 'themeGroup', 'category', 'subtheme']
features = num_cols + cat_cols

pre = ColumnTransformer([
    ('num', Pipeline([('imp', SimpleImputer(strategy='median', add_indicator=True)),
                      ('sc', StandardScaler())]), num_cols),
    ('cat', Pipeline([('imp', SimpleImputer(strategy='constant', fill_value='Unknown')),
                      ('oh', OneHotEncoder(handle_unknown='ignore', min_frequency=5))]), cat_cols),
])

# ANN (preço em escala log, pois é muito assimétrico)
model = Pipeline([('pre', pre), ('ann', TransformedTargetRegressor(
    regressor=MLPRegressor(hidden_layer_sizes=(64, 32), early_stopping=True,
                           max_iter=1000, random_state=42),
    func=np.log1p, inverse_func=np.expm1))])

com_preco = lego_df[~sem_preco]
X_train, X_test, y_train, y_test = train_test_split(
    com_preco[features], com_preco['price'], test_size=0.2, random_state=42)

model.fit(X_train, y_train)
pred = model.predict(X_test)
print(f'MAE: US$ {mean_absolute_error(y_test, pred):.2f} | R²: {r2_score(y_test, pred):.3f}')
```

```python
# Completa os preços faltantes com a ANN
lego_df.loc[sem_preco, 'price'] = model.predict(lego_df.loc[sem_preco, features]).round(2)

print('Preços faltantes em price:', lego_df['price'].isna().sum())   # deve ser 0
lego_df[['name', 'year', 'theme', 'pieces', 'US_retailPrice', 'price']].head(10)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].scatter(y_test, pred, alpha=.4)
ax[0].plot([0, y_test.max()], [0, y_test.max()], 'r--')
ax[0].set(xlabel='Preço real', ylabel='Preço previsto', title='ANN: real vs previsto (teste)')
sns.histplot(x=lego_df['price'], hue=np.where(sem_preco, 'previsto (ANN)', 'original'),
             bins=50, log_scale=True, ax=ax[1])
ax[1].set_title('Distribuição de price')
plt.tight_layout(); plt.show()
```

```python
### Quais colunas são importantes para prever o preço? Importância por permutação: embaralha uma coluna por vez no teste e mede quanto o R² cai.
```

```python
from sklearn.inspection import permutation_importance

r = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42, scoring='r2')
imp = pd.Series(r.importances_mean, index=features).sort_values()
imp.plot.barh(title='Importância das colunas (queda no R²)')
plt.tight_layout(); plt.show()
print(imp.sort_values(ascending=False).round(3))
```
